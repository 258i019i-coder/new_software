#### IUCRの値のゼロ埋め

```python
df["IUCR"].unique()
```

```python
# 4桁にそろえるため，先頭をゼロで埋める
df["IUCR"] = df["IUCR"].str.zfill(4)
print(df["IUCR"].head())
# df.to_csv(rf"Crimes{year}.csv", index = False)
```

#### 変数名の小文字化，スペースのアンダースコアへの変更，不要列の削除

```python
# #変数名のスペースを_に変更

df = df.rename(columns={
    "Case Number": "case_number",
    "Date": "date",
    "Block": "block",
    "Primary Type" : "primary_type",
    "Description" : "description",
    "Location Description" : "location_description",
    "Arrest" : "arrest",
    "Domestic" : "domestic",
    "Community Area" : "community_area",
    "FBI Code" : "fbi_code",
    "Year" : "year"})



df.columns = [
    f"{col.replace(' ', '_')}"
    for col in df.columns
]



df_drop = df.drop(columns = ['X_Coordinate', 'Y_Coordinate', 'Updated_On', 
'Latitude', 'Longitude', 'Location']) 

```

```python
df_drop.columns
```

```python
df_drop.head()
```

```python
# primary_typeの犯罪名を小文字に変換し，スペースをアンダースコア_に変換
df_drop["primary_type"] = (
    df_drop["primary_type"]
      .str.lower()
      .str.replace(r"\s+", "_", regex=True)
)
```

```python
df_drop["primary_type"].unique()
```
